### **Step 3: Start Training**
We will train the `yolo11m.pt` model on your uploaded dataset.

```python
from ultralytics import YOLO
import os

# Fix dataset path in yaml for Colab environment
yaml_path = 'dataset/dataset.yaml'
with open(yaml_path, 'r') as f:
    content = f.read()

# Update path to be absolute content path in Colab
content = content.replace('path: C:\\Users\\talha\\.gemini\\antigravity\\scratch\\malaria_detection\\dataset', 'path: /content/dataset')
content = content.replace('path: ..\\dataset', 'path: /content/dataset')

with open(yaml_path, 'w') as f:
    f.write(content)

# Load Model
model = YOLO('yolo11m.pt')

# Train
results = model.train(
    data=yaml_path,
    epochs=100, # Increased to 100 for better accuracy
    save_period=5, # Save checkpoint every 5 epochs
    imgsz=640,
    batch=16,   # GPU can handle larger batches
    project='malaria_yolo',
    name='cloud_runs',
    exist_ok=True,
    degrees=180.0,
    translate=0.1,
    scale=0.1,
    fliplr=0.5,
    flipud=0.5,
    close_mosaic=10,
    mixup=0.0
)
```
